### 1. Aim
To create a simple Genetic Algorithm by combining selection, crossover, and mutation operations using Python.

### 2. Theory: Putting the GA Cycle Together

A Simple Genetic Algorithm iterates through **Evaluation $\rightarrow$ Selection $\rightarrow$ Crossover $\rightarrow$ Mutation $\rightarrow$ Replacement** until a stopping criterion (e.g., optimal fitness or max generations) is met.

### 3. Step 1: Solving 20-Bit OneMax with Simple GA

```python
import numpy as np

np.random.seed(606)

def run_simple_ga(pop_size=24, chrom_len=20, max_gens=40, pc=0.85, pm=0.04):
    pop = np.random.randint(0, 2, size=(pop_size, chrom_len))
    for gen in range(max_gens):
        fit = np.sum(pop, axis=1)
        best_idx = np.argmax(fit)
        if fit[best_idx] == chrom_len:
            print(f"Converged to optimum at generation {gen}!")
            return pop[best_idx], fit[best_idx]

        next_pop = [pop[best_idx].copy()]
        while len(next_pop) < pop_size:
            c1 = np.random.choice(pop_size, 3, replace=False)
            p1 = pop[c1[np.argmax(fit[c1])]]
            c2 = np.random.choice(pop_size, 3, replace=False)
            p2 = pop[c2[np.argmax(fit[c2])]]

            if np.random.rand() < pc:
                pt = np.random.randint(1, chrom_len)
                o1 = np.concatenate([p1[:pt], p2[pt:]])
                o2 = np.concatenate([p2[:pt], p1[pt:]])
            else:
                o1, o2 = p1.copy(), p2.copy()

            for o in (o1, o2):
                m = np.random.rand(chrom_len) < pm
                o[m] = 1 - o[m]
                if len(next_pop) < pop_size:
                    next_pop.append(o)
        pop = np.array(next_pop)

    fit = np.sum(pop, axis=1)
    return pop[np.argmax(fit)], np.max(fit)

best_sol, best_val = run_simple_ga()
print("Best Solution:", best_sol)
print("Best Fitness: ", best_val)
```

### 4. Conclusion
The complete Simple GA successfully evolved a 20-bit population to the optimal OneMax solution.
