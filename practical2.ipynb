### 1. Aim
To define and calculate a simple fitness function for evaluating individuals in a Genetic Algorithm using Python.

### 2. Theory: Evaluating Candidate Quality

The **fitness function** quantifies how well a chromosome solves the target problem:
- For **maximization** (e.g., OneMax), fitness equals the count of `1`s: $f(\mathbf{x}) = \sum x_i$.
- For **minimization** (e.g., Rastrigin/Sphere), objective cost $g(\mathbf{x})$ is transformed into a positive fitness score: $f(\mathbf{x}) = \frac{1}{1 + g(\mathbf{x})}$.

### 3. Step 1: Computing Fitness for Binary & Continuous Populations

```python
import numpy as np

np.random.seed(202)

def eval_onemax(chrom):
    return int(np.sum(chrom))

def eval_sphere_fitness(chrom):
    cost = np.sum(chrom ** 2)
    return 1.0 / (1.0 + cost)

bin_pop = np.random.randint(0, 2, size=(7, 10))
print("Binary Individuals (OneMax):")
for i, ind in enumerate(bin_pop, 1):
    print(f"  Chrom {i}: {ind} -> Score = {eval_onemax(ind)}")

real_pop = np.round(np.random.uniform(-2.5, 2.5, size=(7, 3)), 2)
print("\nReal-Valued Individuals (Inverse Sphere):")
for i, ind in enumerate(real_pop, 1):
    print(f"  Chrom {i}: {ind} -> Fitness = {eval_sphere_fitness(ind):.4f}")
```

### 4. Step 2: Visualizing Fitness Comparison

```python
import matplotlib.pyplot as plt

plt.style.use('ggplot')
b_scores = [eval_onemax(ind) for ind in bin_pop]
r_scores = [eval_sphere_fitness(ind) for ind in real_pop]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.barh([f"C{i}" for i in range(1, 8)], b_scores, color='#1b998b')
ax1.set_title("OneMax Fitness (Max = 10)")
ax1.set_xlabel("Fitness")

ax2.plot(range(1, 8), r_scores, marker='s', color='#e71d36', linewidth=2)
ax2.set_title("Inverse Sphere Fitness")
ax2.set_xlabel("Individual")
ax2.set_ylabel("Fitness")
plt.tight_layout()
plt.show()
```
