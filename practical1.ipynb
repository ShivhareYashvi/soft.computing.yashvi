### 1. Aim
To create and represent a population of individuals using different encoding methods in Python.

### 2. Theory: Representing Solutions as Chromosomes

Evolutionary algorithms operate on a **population** of candidate solutions, where each candidate is encoded as a **chromosome**. Selecting an appropriate representation directly impacts search efficiency:

- **Binary Encoding:** Uses bitstrings (`0` and `1`), well-suited for subset selection and Boolean optimization.
- **Real-Valued Encoding:** Uses continuous floating-point vectors in $[a, b]$, avoiding discretization error in numerical optimization.
- **Integer Encoding:** Uses discrete whole numbers, ideal for resource allocation and discrete scheduling.
- **Permutation Encoding:** Uses unique ordered sequences of indices, required for routing problems such as TSP.

### 3. Step 1: Generating Populations Across Four Encodings

```python
import numpy as np

np.random.seed(101)

def gen_binary_pop(n_ind, length):
    return np.random.randint(0, 2, size=(n_ind, length))

def gen_real_pop(n_ind, length, low=-10.0, high=10.0):
    return np.round(np.random.uniform(low, high, size=(n_ind, length)), 2)

def gen_int_pop(n_ind, length, low=1, high=15):
    return np.random.randint(low, high + 1, size=(n_ind, length))

def gen_perm_pop(n_ind, length):
    return np.array([np.random.permutation(length) for _ in range(n_ind)])

n_ind, length = 4, 8
print("Binary Population:\n", gen_binary_pop(n_ind, length))
print("\nReal-Valued Population:\n", gen_real_pop(n_ind, length))
print("\nInteger Population:\n", gen_int_pop(n_ind, length))
print("\nPermutation Population:\n", gen_perm_pop(n_ind, length))
```

### 4. Step 2: Visualizing Gene Distributions

```python
import matplotlib.pyplot as plt

plt.style.use('ggplot')
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

b_pop = gen_binary_pop(60, 12)
axes[0].barh(['Bit 0', 'Bit 1'], [np.sum(b_pop == 0), np.sum(b_pop == 1)], color=['#2b5c8f', '#e07a5f'])
axes[0].set_title("Binary Bit Frequency")
axes[0].set_xlabel("Count")

r_pop = gen_real_pop(120, 4)
axes[1].hist(r_pop.flatten(), bins=12, color='#3d9970', edgecolor='white', alpha=0.85)
axes[1].set_title("Real-Valued Gene Histogram")
axes[1].set_xlabel("Gene Value")

i_pop = gen_int_pop(100, 4, 1, 10)
u_vals, u_cnts = np.unique(i_pop, return_counts=True)
axes[2].stem(u_vals, u_cnts, linefmt='#6a4c93', markerfmt='o', basefmt=' ')
axes[2].set_title("Integer Gene Stem Plot")
axes[2].set_xlabel("Integer")

plt.tight_layout()
plt.show()
```
