### 2. Theory: Maintaining Genetic Diversity

Mutation introduces stochastic perturbations so the population does not stagnate in local optima:
- **Bit-Flip Mutation:** Inverts binary genes with probability $p_m$.
- **Gaussian Mutation:** Perturbs real-valued genes with $\mathcal{N}(0, \sigma^2)$ noise.
- **Swap Mutation:** Exchanges two randomly selected positions in a permutation.

### 3. Step 1: Applying Bit-Flip, Gaussian, and Swap Mutation

```python
import numpy as np

np.random.seed(505)

def mutate_bitflip(ind, pm=0.2):
    out = ind.copy()
    flips = np.random.rand(len(out)) < pm
    out[flips] = 1 - out[flips]
    return out

def mutate_gaussian(ind, pm=0.35, sigma=0.4):
    out = ind.astype(float).copy()
    mask = np.random.rand(len(out)) < pm
    out[mask] += np.random.normal(0, sigma, size=np.sum(mask))
    return np.round(out, 3)

def mutate_swap(ind):
    out = ind.copy()
    i, j = np.random.choice(len(out), 2, replace=False)
    out[i], out[j] = out[j], out[i]
    return out, (i, j)

b = np.array([0, 1, 0, 1, 1, 0, 0, 1])
print("Binary Before:", b)
print("Binary After: ", mutate_bitflip(b))

r = np.array([2.5, -1.0, 0.4, 3.2, -2.1])
print("\nReal Before:  ", r)
print("Real After:   ", mutate_gaussian(r))

p = np.array([0, 1, 2, 3, 4, 5, 6])
pm_out, pair = mutate_swap(p)
print("\nPerm Before:  ", p)
print(f"Perm After (swapped {pair}):", pm_out)
```

### 4. Conclusion
Bit-Flip, Gaussian, and Swap mutation operators were implemented to introduce controlled variation across different encodings.
